# **Advanced NLP Techniques**
In order for us to have a comprehensive overview of advanced techniques in Natural Language Processing (NLP), we may highlight the evolution of language models from traditional approaches, such as Bag of Words, to deep learning-based models like Word2Vec and BERT. These technologies have improved the semantic and contextual representation of language, with BERT introducing dynamic vectors and bidirectional self-supervised training, making it highly effective in tasks such as text classification, sentiment analysis, and entity recognition. It's important to consider the value of pre-trained models and their variants, such as RoBERTa, DistilBERT, and ALBERT, which balance performance and efficiency.

```python
from transformers import pipeline, BertTokenizer, BertForSequenceClassification, Trainer, TrainingArguments
```

## **Example: BERT usage**
This example shows how BERT can predict words that make sense in the given context, demonstrating its ability to understand and generate language in a coherent and relevant way.

```python
# Creating a pipeline 'fill-mask' using BERT:
fill_mask = pipeline ('fill-mask', model='bert-base-uncased')

# Exemplo de preenchimento de lacuna
result = fill_mask('The new technology is [MASK] to change the world.')
print([r['sequence'] for r in result])
```

## **Example: Using BERTimbau in a fill-in-the-blank task**
The following code uses the Transformers library to load the BERTimbau model and apply it to a gap-filling task, demonstrating its effectiveness in understanding and generating text in Portuguese.

```python
# Initialize a gap-filling pipeline using Bertimbau:
fill_mask = pipeline ('fill-mask', model='neuralmind/bert-base-portuguese-cased')

# Example of fill-in-the-blank:
result = fill_mask('O clima em Lisboa é geralmente [MASK].')
print([r['sequence'] for r in result])
```

## **Example: Fine Tuning code**
The following code illustrates how to load and prepare a BERTimbau model for fine-tuning in a specific text classification task, using the Transformers library.

```python
from torch.utils.data import Dataset

# Loading the tokenizer and the model:
tokenizer = BertTokenizer.from_pretrained('neuralmind/bert-base-portuguese-cased')
model = BertForSequenceClassification.from_pretrained('neuralmind/bert-base-portuguese-cased', num_labels=3)

# Preparing the training data:
# NOTE: 'train_texts', 'train_labels', and 'Dataset' are not defined/imported in this context.
# You would typically define 'train_texts' and 'train_labels' and import 'Dataset' from 'torch.utils.data' or 'datasets'.

# Dummy data for demonstration purposes:
train_texts = ["Isso é um texto de exemplo.", "Outro exemplo de texto.", "Mais um texto para treinamento."]
train_labels = [0, 1, 0]

train_encodings = tokenizer(train_texts, padding=True, max_length=128, truncation=True)

class CustomDataset(Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels

    def __getitem__(self, idx):
        item = {key: val[idx] for key, val in self.encodings.items()}
        item['labels'] = self.labels[idx]
        return item

    def __len__(self):
        return len(self.labels)

train_dataset = CustomDataset(train_encodings, train_labels)

# Defining the training arguments:
training_args = TrainingArguments(output_dir='./results', num_train_epochs=3, per_device_train_batch_size=16)

# Training the model:
trainer = Trainer(model =model, args=training_args, train_dataset=train_dataset)
trainer.train()
```
