## **Example: Tokenization and Normalization**
To illustrate, we can consider the following example that performs tokenization and normalization of a text.

```python
# Text example:
text = 'Tokenization and normalization are foundational preprocessing steps in natural language processing (NLP) that transform raw, unstructured text into a structured, consistent format suitable for machine learning models.'

# Tokenization:
tokens = word_tokenize(text)

# Normalization: converting to capital letters:
normalized_tokens = [token.lower() for token in tokens]
print('Tokens:', tokens)

print('Normalized tokens:', normalized_tokens)
```

## **Example: Implementation of Preprocessing, Tokenization, and Normalization**

Implementing preprocessing, tokenization, and normalization of text is a crucial step in the analysis of textual data, especially in natural language. Let's exemplify this process using Python and a subtitle file. We begin by downloading the subtitle file, which has a specific structure with sequence numbers, timestamps, and the spoken text.

```python
# Assuming the subtitle file has already been downloaded and is available as 'file.txt':
with open('Hatari.English-WWW.MY-SUBS.CO.srt', 'r') as file:
  raw_subtitles = file.read()

# Simplification and cleaning of the text:
clean_subtitles = raw_subtitles.split('\n\n')
clean_subtitles = [text.split('\n')[-1] for text in clean_subtitles]

# Putting it all together in a single text:
full_text = ' '.join(clean_subtitles)

# Tokenization and normalization:
sentences = full_text.split('. ')
tokens = [word.lower() for sentence in sentences for word in sentence.split()]

# Displaying the first tokens for verification:
print(tokens[:10])
```

```python
# Let's assume that 'text' is a string containing the text from which we want to extract keywords:
text = 'Tokenization and normalization are foundational preprocessing steps in natural language processing (NLP) that transform raw, unstructured text into a structured, consistent format suitable for machine learning models.'

# Tokenization and removal of stop words.
nltk.download('punkt')
nltk.download('stopwords')
tokens = word_tokenize(text)
clean_tokens = [word for word in tokens if word.lower() not in stopwords.words('english')]

# Calculating word frequency:
freq_dist = FreqDist(clean_tokens)
keywords = freq_dist.most_common(5)   # Gets the 5 most frequent words.
print(keywords)
```

```python
text = 'Tokenization and normalization are foundational preprocessing steps in natural language processing (NLP) that transform raw, unstructured text into a structured, consistent format suitable for machine learning models.'

kw_extractor = yake.KeywordExtractor(lan='en', n=1)
keywords = kw_extractor.extract_keywords(text)
print(keywords)
```
