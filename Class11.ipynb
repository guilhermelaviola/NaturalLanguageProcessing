# **Sentiment Analysis**
Sentiment analysis is a fundamental area of ​​Natural Language Processing that seeks to identify and classify emotions expressed in texts, going beyond positive or negative polarity to capture more complex sentiments. Widely applied in corporate and socio-political contexts, it allows for the assessment of public perception, the identification of opinion trends, and the support of strategic decision-making. This analysis can be performed at different levels of granularity and through rule-based approaches, machine learning, or hybrid models, each with its advantages and challenges. Tools like TextBlob facilitate its implementation, although obstacles such as sarcasm, irony, and figurative language still require careful pre-processing. With the continuous evolution of models and algorithms, sentiment analysis tends to become increasingly precise and indispensable for extracting relevant insights from textual data in various fields.

```python
# Importing all the necessary libraries and resources:
from textblob import TextBlob
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
import re
from textblob import TextBlob
from datasets import load_dataset
```

```python
# Exemplo simplificado de código Python para análise de sentiments usando a biblioteca TextBlob
# text de exemplo
text = 'O hotel era maravilhoso!'

# Análise de sentiments
analysis = TextBlob(text)
polarity = analysis.sentiment.polarity  # Valor entre -1 e 1, onde valores mais altos indicam sentiments positives

# Saída de resultados
if polarity > 0:
  print("Positive")
elif polarity == 0:
  print("Neutral")
else:
  print("Negative")
```

```python
# Exemplo de uma regra simples em Python para classificação de sentiments
def classify_sentiment(text):
  positives = ["excellent", "very good", "fantastic"]
  negatives = ["bad", "awful", "terrible"]

  for word in text.split():
    if word in positives:
      return "Positive"
    elif word in negatives:
      return "Negative"

  return "Neutral"

# Teste da função
print(classify_sentiment("The movie was excellent"))
```

```python
# Exemplo simplificado de um model de classificação de sentiments usando Scikit-Learn
# Dados de treinamento
texts = ["The movie was excellent", "The menu was terrible"]
sentiments = [1, 0]  # 1 para positivo, 0 para negativo

# Transformação dos texts em vetores
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)

# Treinamento do model
model = LogisticRegression()
model.fit(X, sentiments)

# Previsão de um novo text
new_text = vectorizer.transform(["The work was really good"])
print("Prediction:", "Positive" if model.predict(new_text)[0] == 1 else "Negative")
```

```python
# Exemplo de código Python para pré-processamento de text
def pre_processar_text(text):

  # Corrigir erros comuns de ortografia
  text = re.sub(r"excellent", "excellent", text)

  # Tratar sarcasmo identificando padrões comuns
  sarcasm_pattern = re.compile(r"(tão bom que dormi|bom se você tiver 5 anos)", re.IGNORECASE)
  if sarcasm_pattern.search(text):
    return "Negative"

  # Remover emojis e converter abreviações
  text = re.sub(r":)', "", text)
  text = text.replace("y", "you")
  return text

# Exemplo de aplicação do pré-processamento
text_review = "O filme era tão bom que eu dormi. Excelnte!"
processed_text = pre_process_text(text_review)
print(processed_text)
```

```python
# Carregando o dataset
dataset = load_dataset("huggingface_dataset_name", "subset_name")

# Exemplo de acesso aos dados
print(dataset["train"][0]["review_text"])  # Exibe o text da primeira avaliação
```

```python
for review in dataset["train"]:
  text = review["review_text"]

  if text:  # Verifica se o text da avaliação não está vazio
  sentiment = TextBlob(text).sentiment.polarity
  print(f"text: {text}\npolarity: {sentiment}\nEstrelas: {review["rating"]}")
```

```python
def analyze_sentiments(reviews):
  results = []

  for review in reviews:
    if review["review_text"]:  # Certifique-se que a avaliação contém text
    polarity = TextBlob(review["review_text"]).sentiment.polarity
    stars = review["rating"]
    results.append((stars, polarity))

  return results

# Análise de todas as avaliações
all_reviews = analyze_sentiments(dataset["train"])

# Exibe algumas das análises
print(all_reviews[:10])
```

```python
strange_analyses = [review for review in all_reviews if (review[0] >= 4 and review[1] < 0.2) or (review[0] <= 2 and review[1] > 0.2)]
print(f"Número de análises estranhas: {len(strange_analyses)}")
```
