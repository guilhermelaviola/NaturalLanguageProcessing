## **Example: Using Leading Generative Models**

Generative models, such as GPT-4 and similar ones, are initially trained on vast datasets of text. After this basic training, they undergo a fine-tuning process, which specializes them to perform specific tasks according to the instructions they receive. This allows these models not only to generate text but also to retain the context of an interaction, enabling long and meaningful conversations with users. To implement ChatGPT in a program, the process is relatively simple with the help of the OpenAI library. The following code demonstrates how to start an instance of the model and configure it to complete a specific task.

```python
# Inictializing the OpenAI instance:
import os
# IMPORTANT: Replace 'YOUR_API_KEY' with your actual OpenAI API key
# For security, consider loading this from environment variables or a secure secret manager.
os.environ['OPENAI_API_KEY'] = 'YOUR_API_KEY'
openai_api = openai.OpenAI()

# Creating a class for the model:
class ChatGPT:
  def __init__(self, model_name='gpt-3.5-turbo'):
    self.model = model_name # Fixed: Assign model_name to self.model

  def generate_text(self, prompt):
    # Using chat completions for gpt-3.5-turbo model
    completion = openai_api.chat.completions.create(
        model=self.model, # Fixed: Used 'model' parameter and '=' for assignment
        messages=[
            {"role": "system", "content": "You are a poetic assistant; explain the concept of recursion through a poem."}, # Added a system message for context
            {"role": "user", "content": prompt}
        ],
        max_tokens=150
    )
    return completion.choices[0].message.content.strip() # Accessing content from message object

# Using the model to generate a poem about recursion:
assistant = ChatGPT()
prompt = 'Explain the concept of recursion through a poem.' # Simplified prompt as system message provides context
print(assistant.generate_text(prompt))
```
