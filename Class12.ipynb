# **Machine Learning & Deep Learning for Natural Language**
Natural language processing has been transformed by machine learning and deep learning, enabling computers to understand and generate human language for tasks such as translation, sentiment analysis, and summarization. Machine learning allows models to learn patterns from large amounts of data, while deep learning uses multi-layer neural networks to automatically extract complex linguistic features through techniques like embeddings. Advances in architectures such as CNNs, RNNs, and especially Transformers have led to powerful large language models like BERT and GPT, which can be fine-tuned for many applications with relatively little labeled data. Despite these advances, challenges remain, including the need for high-quality labeled data, limited model interpretability, and ethical concerns such as bias and fairness. Open-source tools have made NLP development more accessible, but responsible design and deployment are essential to ensure these technologies are transparent, ethical, and beneficial to society.

```python
# Importing all the necessary libraries and resources:
import pandas as pd
from datasets import load_dataset
from transformers import pipeline
```

```python
# Loading the dataset and displaying the reviews and stars:
url = 'https://raw.githubusercontent.com/AFAgarap/ecommerce-reviews-analysis/refs/heads/master/Womens%20Clothing%20E-Commerce%20Reviews.csv'
df = pd.read_csv(url)

# Accessing 'Review Text' column directly from the DataFrame:
reviews = df['Review Text']
print(f'First review: {reviews[0]}')

# Accessing 'Rating' column directly from the DataFrame:
stars = df['Rating']
print(f'First star rating: {stars[0]}')

print(f'Total number of reviews: {len(reviews)}')
```

```python
classifier = pipeline(
    model='lxyuan/distilbert-base-multilingual-cased-sentiments-student',
    return_all_scores=True,
)

classifier('The movie was excellent!')
classifier('The movie was terrible!')
classifier('The movie was Okay!')
classifier('I cannot say if I liked the movie or not.')
classifier(['I cannot say I liked the movie', 'I loved the movie'])

print([x for x in reviews if not x])
reviews = [x if x else '' for x in reviews]
print([x for x in reviews if not x])

# For all the reviews, the classification would take around 1 hour:
classifications = classifier(reviews[:100])

for i, review in enumerate(reviews[:100]):
  if review:
    print((stars[i], classifications[i]))
```
