# **Machine Learning & Deep Learning for Natural Language**
Natural language processing has been transformed by machine learning and deep learning, enabling computers to understand and generate human language for tasks such as translation, sentiment analysis, and text generation. Machine learning allows models to learn patterns from large text datasets, while deep learning uses multi-layer neural networks to automatically extract complex features, with embeddings capturing semantic relationships between words. Advances in neural architectures—including CNNs, RNNs, and especially Transformers—along with large language models like BERT and GPT, have significantly improved performance and adaptability across NLP tasks. Despite these advances, challenges remain, including the need for large, high-quality labeled datasets, limited model interpretability, and ethical concerns such as bias and fairness. Open-source tools have made NLP development more accessible, but responsible deployment requires careful attention to transparency, equity, and societal impact.

```python
# importing all the necessary libraries and resources:
import pandas as pd
from datasets import load_dataset
from transformers import pipeline
```

## **Sentiment Analysis Model Using Neural Networks**
In this part of our Natural Language Processing (NLP) course, we will apply Deep Learning concepts to build a sentiment analysis model using deep neural networks. We will use a pre-trained model based on DistilBERT, a lighter and faster version of BERT, developed by Google.

```python
# Reading a dataset from URL:
url = 'https://raw.githubusercontent.com/AFAgarap/ecommerce-reviews-analysis/refs/heads/master/Womens%20Clothing%20E-Commerce%20Reviews.csv'
dataset = pd.read_csv(url)
dataset

reviews = dataset['Review Text']
# In the original code, 'dataset' was loaded as a DataFrame, not a Dataset object with 'train' key.
# Accessing columns directly from the DataFrame 'dataset' will be correct.
# reviews = dataset['train']['Review Text'] is incorrect, it should be reviews = dataset['Review Text']
reviews[0]

stars = dataset['Rating']
# stars = dataset['train']['Rating'] is incorrect, it should be stars = dataset['Rating']
stars[0]

len(reviews)
len(stars)
```

```python
# Defining the classifier:
classifier = pipeline(
    model='lxyuan/distilbert-base-multilingual-cased-sentiments-student',
    return_all_scores=True,
)

classifier('I think the movie was excellent!')
classifier('I think the movie was terrible!')
classifier('I think the movie was okay')
classifier('I cannot say if I liked the movie')
classifier(['I do not know if I liked the movie', 'I loved the movie'])

print([x for x in reviews if not x])

reviews = [x if x else '' for x in reviews]
print([x for x in reviews if not x])
```

```python
# For all the reviews (over 23.000), the classification would take a while.
cleaned_reviews = ['' if pd.isna(r) else str(r) for r in reviews[:100]]

classifications = classifier(cleaned_reviews)

for i, review in enumerate(cleaned_reviews):
    if review:
        print((stars[i], classifications[i]))
```
